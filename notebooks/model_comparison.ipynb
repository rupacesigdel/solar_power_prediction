```python
import matplotlib.pyplot as plt
import pandas as pd

results = pd.DataFrame({
    'Model': ['ANN', 'LSTM'],
    'MAE': [0.0705861217307695, 0.052872574418216585],
    'RMSE': [0.11995842499367056, 0.0905811303028474],
    'R2': [0.8286530302448556, 0.9002770763501181],
    'MAPE': [21.95322908963514, 14.760286928033093]
})

results
```

```python
# RMSE Comparison
plt.figure(figsize=(6,5))

plt.bar(results['Model'], results['RMSE'])

plt.title("RMSE Comparison")

plt.ylabel("RMSE")

plt.show()
```

```python
# R^2 Comparison
plt.figure(figsize=(6,5))

plt.bar(results['Model'], results['R2'])

plt.title("R2 Score Comparison")

plt.ylabel("R2")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.bar(results['Model'], results['MAPE'])

plt.title("MAPE Comparison")

plt.ylabel("MAPE")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.bar(results['Model'], results['MAE'])

plt.title("MAE Comparison")

plt.ylabel("MAE")

plt.show()
```
